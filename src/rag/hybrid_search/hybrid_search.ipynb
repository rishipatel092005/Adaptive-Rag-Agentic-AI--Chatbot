```python
from pathlib import Path
from dotenv import load_dotenv
import os
import json
import re

import numpy as np
import pandas as pd

# --------------------------------------------------
# 1. Find project root and load .env
# --------------------------------------------------

CURRENT_DIR = Path.cwd()

# Look for .env in current directory or parent directories
ENV_PATH = next(
    (
        path / ".env"
        for path in [CURRENT_DIR, *CURRENT_DIR.parents]
        if (path / ".env").exists()
    ),
    None
)

if ENV_PATH is None:
    raise FileNotFoundError(
        "Could not find .env file in the notebook folder or any parent folder."
    )

load_dotenv(ENV_PATH, override=True)

print("Environment file:", ENV_PATH)


# --------------------------------------------------
# 2. Groq API key
# --------------------------------------------------

GROQ_API_KEY = os.getenv("GROQ_API_KEY")

print(
    "Groq API key loaded:",
    bool(GROQ_API_KEY)
)


# --------------------------------------------------
# 3. Dataset path
# --------------------------------------------------

DATA_PATH = CURRENT_DIR / "products_hybrid_rag_200.json"

print("Dataset path:", DATA_PATH)
print("Dataset exists:", DATA_PATH.exists())


# --------------------------------------------------
# 4. Load dataset
# --------------------------------------------------

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATA_PATH}"
    )

with open(DATA_PATH, "r", encoding="utf-8") as f:
    products = json.load(f)

print("Number of products:", len(products))


# --------------------------------------------------
# 5. Basic dataset check
# --------------------------------------------------

required_fields = {
    "sku",
    "category",
    "name",
    "description"
}

missing_fields = required_fields - set(products[0].keys())

if missing_fields:
    raise ValueError(
        f"Missing fields in dataset: {missing_fields}"
    )

print("Dataset structure: OK")
print("First product:")
print(products[0])
```
